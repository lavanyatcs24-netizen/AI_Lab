```python
# Vacuum Cleaner - Goal Based Agent
# State Space Representation

# Initial state
# (Vacuum Position, Room A, Room B)
state = ['A', 'Dirty', 'Dirty']

# Goal state
goal = ['A', 'Clean', 'Clean']


def display_state():
    print("Vacuum Position :", state[0])
    print("Room A          :", state[1])
    print("Room B          :", state[2])
    print()


print("Initial State")
display_state()

# Continue until goal state is reached
while state[1] != 'Clean' or state[2] != 'Clean':

    # If current room is dirty -> Suck
    if state[0] == 'A' and state[1] == 'Dirty':
        print("Action: S")
        state[1] = 'Clean'

    elif state[0] == 'B' and state[2] == 'Dirty':
        print("Action: S")
        state[2] = 'Clean'

    # If A is clean but B is dirty -> Move Right
    elif state[0] == 'A' and state[1] == 'Clean' and state[2] == 'Dirty':
        print("Action: R")
        state[0] = 'B'

    # If B is clean but A is dirty -> Move Left
    elif state[0] == 'B' and state[2] == 'Clean' and state[1] == 'Dirty':
        print("Action: L")
        state[0] = 'A'


print("\nGoal State Reached")
display_state()
```
