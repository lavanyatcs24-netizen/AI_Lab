```python
import math
import random


def get_cost(board):
    """Calculate the number of attacking pairs of queens."""
    attacks = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            if (
                board[i] == board[j]
                or abs(board[i] - board[j]) == abs(i - j)
            ):
                attacks += 1
    return attacks


def get_random_neighbor(board):
    """Generate a neighbor by moving one queen to a random row."""
    neighbor = list(board)
    col = random.randint(0, len(board) - 1)
    new_row = random.randint(0, len(board) - 1)
    while neighbor[col] == new_row:
        new_row = random.randint(0, len(board) - 1)
    neighbor[col] = new_row
    return neighbor


def simulated_annealing(
    n=8, init_temp=1000.0, cooling_rate=0.995, min_temp=1e-8, max_iterations=100000
):
    current_board = [random.randint(0, n - 1) for _ in range(n)]
    current_cost = get_cost(current_board)
    temp = init_temp

    # Track global minimum and best state found during execution
    best_board = list(current_board)
    best_cost = current_cost
    total_iterations = 0

    for step in range(max_iterations):
        total_iterations += 1

        if current_cost == 0:
            return current_board, current_cost, total_iterations, best_board, best_cost

        if temp < min_temp:
            break

        neighbor = get_random_neighbor(current_board)
        neighbor_cost = get_cost(neighbor)
        cost_diff = neighbor_cost - current_cost

        # Acceptance check
        if cost_diff < 0 or random.uniform(0, 1) < math.exp(-cost_diff / temp):
            current_board = neighbor
            current_cost = neighbor_cost

            # Keep record of the global best state
            if current_cost < best_cost:
                best_cost = current_cost
                best_board = list(current_board)

        temp *= cooling_rate

    return current_board, current_cost, total_iterations, best_board, best_cost


def print_board(board):
    """Print the board configuration visually with Queens."""
    n = len(board)
    for row in range(n):
        row_str = []
        for col in range(n):
            if board[col] == row:
                row_str.append("♛")
            else:
                row_str.append("·")
        print(" ".join(row_str))


if __name__ == "__main__":
    # Seed chosen to show tracking mechanics
    random.seed(50)

    final_board, final_cost, total_iters, best_board, best_cost = simulated_annealing()

    print(f"Total Iterations Tracked: {total_iters}")
    print(f"Minimum Conflicts (Best Cost) Found: {best_cost}\n")
    print("--- Best Solution Board View ---")
    print_board(best_board)
```
