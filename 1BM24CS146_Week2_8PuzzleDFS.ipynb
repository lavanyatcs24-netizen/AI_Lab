```python
GOAL = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

MOVES = {
    0: [1, 3],
    1: [0, 2, 4],
    2: [1, 5],
    3: [0, 4, 6],
    4: [1, 3, 5, 7],
    5: [2, 4, 8],
    6: [3, 7],
    7: [4, 6, 8],
    8: [5, 7]
}


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def get_neighbors(state):
    neighbors = []

    blank = state.index(0)

    for move in MOVES[blank]:
        new_state = list(state)

        new_state[blank], new_state[move] = \
            new_state[move], new_state[blank]

        neighbors.append(tuple(new_state))

    return neighbors


def dfs(start):
    stack = [(start, [start])]
    visited = set()

    while stack:

        state, path = stack.pop()

        if state in visited:
            continue

        visited.add(state)

        if state == GOAL:
            return path

        for neighbor in get_neighbors(state):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))

    return None


start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)

print("1BM24CS146 LAVANYA T")
print("INITIAL STATE:")
print_puzzle(start)

print("GOAL STATE:")
print_puzzle(GOAL)

solution = dfs(start)

if solution:
    print("DFS SOLUTION:")
    print("Number of moves:", len(solution) - 1)
    print()

    for i, state in enumerate(solution):
        print("Step", i)
        print_puzzle(state)
else:
    print("No solution found.")
```
