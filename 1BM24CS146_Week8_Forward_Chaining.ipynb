```python

# First Order Logic - Forward Chaining
# Problem: Robert is Criminal
# Takes complete English sentences as input

import re

def forward_chaining(text):

    text = text.lower()

    # Initial Knowledge Base
    KB = set()

    print("\nProcessing the given statements...")

    # Extract facts from user input

    if "american" in text and "robert" in text:
        KB.add("American(Robert)")

    if "enemy of america" in text or "enemy of american" in text:
        KB.add("Enemy(A,America)")

    if "missile" in text and "country a" in text:
        KB.add("Missile(T1)")
        KB.add("Owns(A,T1)")

    if ("sold" in text or "sell" in text) and "robert" in text:
        KB.add("Sells(Robert,T1,A)")

    # Display initial facts
    print("\nInitial Knowledge Base:")
    for fact in sorted(KB):
        print(fact)

    # Forward Chaining
    while True:

        new_facts = set()

        # Rule 1: Enemy -> Hostile
        if "Enemy(A,America)" in KB:
            new_facts.add("Hostile(A)")

        # Rule 2: Missile -> Weapon
        if "Missile(T1)" in KB:
            new_facts.add("Weapon(T1)")

        # Rule 3: American AND Weapon AND Sells AND Hostile
        # -> Criminal
        if (
            "American(Robert)" in KB
            and "Weapon(T1)" in KB
            and "Sells(Robert,T1,A)" in KB
            and "Hostile(A)" in KB
        ):
            new_facts.add("Criminal(Robert)")

        # Remove already existing facts
        new_facts = new_facts - KB

        # Stop if no new facts
        if not new_facts:
            break

        # Add inferred facts
        KB.update(new_facts)

        print("\nNewly Inferred Facts:")
        for fact in sorted(new_facts):
            print(fact)

    # Check Goal
    print("\nFinal Result:")

    if "Criminal(Robert)" in KB:
        print("GOAL PROVED!")
        print("Robert is a Criminal.")
    else:
        print("GOAL NOT PROVED!")
        print("Robert cannot be proved as a Criminal.")

# Get complete problem statement
print("FORWARD CHAINING - ROBERT IS CRIMINAL")
print("--------------------------------------")

print("\nEnter the complete problem statement:")
text = input()

# Execute algorithm
forward_chaining(text)
```
