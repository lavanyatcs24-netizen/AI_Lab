```python
import random

# Create the board
board = ['1', '2', '3',
         '4', '5', '6',
         '7', '8', '9']


# Display the board
def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


# Check whether a player has won
def check_winner(symbol):
    winning_combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_combinations:
        if board[a] == symbol and board[b] == symbol and board[c] == symbol:
            return True

    return False


# Check whether the board is full
def board_full():
    for cell in board:
        if cell not in ['X', 'O']:
            return False
    return True


# Main game
print("TIC-TAC-TOE")
print("Human = X")
print("Computer = O")

while True:

    display_board()

    # Human move
    while True:
        try:
            position = int(input("Enter your position (1-9): "))

            if position < 1 or position > 9:
                print("Enter a number between 1 and 9.")
            elif board[position - 1] in ['X', 'O']:
                print("Position already occupied. Try again.")
            else:
                board[position - 1] = 'X'
                break

        except ValueError:
            print("Please enter a valid number.")

    # Check human win
    if check_winner('X'):
        display_board()
        print("Human wins!")
        break

    # Check draw
    if board_full():
        display_board()
        print("Game is a draw!")
        break

    # Computer move
    available_positions = []

    for i in range(9):
        if board[i] not in ['X', 'O']:
            available_positions.append(i)

    computer_position = random.choice(available_positions)
    board[computer_position] = 'O'

    print("Computer selected position:", computer_position + 1)

    # Check computer win
    if check_winner('O'):
        display_board()
        print("Computer wins!")
        break

    # Check draw
    if board_full():
        display_board()
        print("Game is a draw!")
```

```python
# Vacuum Cleaner - Goal Based Agent
# State Space Representation

# Initial state
# (Vacuum Position, Room A, Room B)
state = ['A', 'Dirty', 'Dirty']

# Goal state
goal = ['A', 'Clean', 'Clean']


def display_state():
    print("Vacuum Position :", state[0])
    print("Room A          :", state[1])
    print("Room B          :", state[2])
    print()


print("Initial State")
display_state()

# Continue until goal state is reached
while state[1] != 'Clean' or state[2] != 'Clean':

    # If current room is dirty -> Suck
    if state[0] == 'A' and state[1] == 'Dirty':
        print("Action: S")
        state[1] = 'Clean'

    elif state[0] == 'B' and state[2] == 'Dirty':
        print("Action: S")
        state[2] = 'Clean'

    # If A is clean but B is dirty -> Move Right
    elif state[0] == 'A' and state[1] == 'Clean' and state[2] == 'Dirty':
        print("Action: R")
        state[0] = 'B'

    # If B is clean but A is dirty -> Move Left
    elif state[0] == 'B' and state[2] == 'Clean' and state[1] == 'Dirty':
        print("Action: L")
        state[0] = 'A'


print("\nGoal State Reached")
display_state()
```
