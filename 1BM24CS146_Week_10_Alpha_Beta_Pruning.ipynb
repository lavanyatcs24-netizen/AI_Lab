```python
import math

# 1. ALPHA-BETA SEARCH
def alpha_beta(node, alpha, beta, is_max, path="ROOT", pruned=None):
    if pruned is None:
        pruned = []

    # If we reached a leaf number at the bottom
    if isinstance(node, int):
        print(f"Visited: {path} = {node}")
        return node

    if is_max:
        v = -math.inf
        for i, child in enumerate(node):
            v = max(v, alpha_beta(child, alpha, beta, False, f"{path}->B{i+1}", pruned))
            if v >= beta:
                print(f"  [PRUNED at {path}] (v={v} >= beta={beta})")
                for rem in node[i+1:]:
                    collect_leaves(rem, pruned)
                return v
            alpha = max(alpha, v)
        return v
    else:
        v = math.inf
        for i, child in enumerate(node):
            v = min(v, alpha_beta(child, alpha, beta, True, f"{path}->B{i+1}", pruned))
            if v <= alpha:
                print(f"  [PRUNED at {path}] (v={v} <= alpha={alpha})")
                for rem in node[i+1:]:
                    collect_leaves(rem, pruned)
                return v
            beta = min(beta, v)
        return v

def collect_leaves(node, pruned):
    if isinstance(node, int):
        pruned.append(node)
        return
    for child in node:
        collect_leaves(child, pruned)

# 2. SIMPLE INPUT: Enter any tree using brackets [ ]
print("Paste your tree using brackets [ ] and press Enter:")
tree = eval(input("Tree: "))

pruned_nodes = []
ans = alpha_beta(tree, -math.inf, math.inf, True, pruned=pruned_nodes)

print("\n" + "="*30)
print("Root Value   :", ans)
print("Pruned Leaves:", pruned_nodes)
print("="*30)
```
